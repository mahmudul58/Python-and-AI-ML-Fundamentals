```python
import pandas as pd

df=pd.read_csv("./Practice Day.csv")
df
```

### **Problem 1: Add New Column and Save File**

#### **Tasks:**
* You have a DataFrame with column `Price`.
* Create a new column `Discounted_Price` with a **10% discounted price**.
* Save the modified DataFrame as `discounted_products.csv`.

```python
df['Discounted_Price']=df['Price']*.9
df.to_csv("discounted_products.csv")
```

### **Problem 2: Check Duplicates and Remove Them**

**Column:** `EmployeeID`

#### **Tasks:**
1. Count duplicate entries.
2. Remove duplicates.
3. Display the cleaned DataFrame.

```python
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
display(df)
```

### **Problem 3: Handle Missing Values**

**Columns:** `Age`, `Salary`

#### **Tasks:**
1. Count missing values in these columns.
2. Fill missing `Age` values with the **mean age**.
3. Drop rows where `Salary` is missing.

```python
print(f"Missing Values in Age Column: {df['Age'].isnull().sum()}")

df['Age']=df['Age'].fillna(df['Age'].mean())
display(df)
print(f"Missing Values in Salary Column: {df['Salary'].isnull().sum()}")
df.dropna(inplace=True)
```

### **Problem 4: Apply Statistical Functions**

**Columns:** `Marks1`, `Marks2`, `Marks3`

#### **Tasks:**
1. Calculate the **mean** and **max** for each subject.
2. Compute the **total marks** per student.
3. Find the **standard deviation** of total marks.

```python
print(f"Mean of \"Marks1\": {df['Marks1'].mean().round(2)} and Max :{df['Marks1'].max()}\n")
print(f"Mean of \"Marks2\": {df['Marks2'].mean().round(2)} and Max :{df['Marks2'].max()}\n")
print(f"Mean of \"Marks3\": {df['Marks3'].mean().round(2)} and Max :{df['Marks3'].max()}\n")

df['Total Marks']=df['Marks1']+df['Marks2']+df['Marks3']
df
```
