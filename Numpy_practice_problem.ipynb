**1.** You have a list of temperatures recorded in Celsius. Your task is to:  
* Create a NumPy array from the given list.  
* Convert all temperatures from Celsius to Fahrenheit. The formula is $F = C \times 1.8 + 32$.  
* Print the new Fahrenheit array.  



```python
import numpy as np

temp_in_celsius=[10,20,-40,50,-100,273]

temp_in_fahrenheit=(np.array(temp_in_celsius))*1.8+32

print(temp_in_fahrenheit)

```

**2.** You have a Python list containing the lap times (in seconds) for an athlete.  
* **Task:**  
  * Convert the list into a NumPy array.  
  * Print the array's shape, size, and data type.  
  * The athlete wants to see their times in minutes. Create a new array by dividing all the lap times by 60.  
  * Print the new array of lap times in minutes.  
</div>

```python
import numpy as np

lap_times = [64.2, 60, 65.5, 59.3]

arr=np.array(lap_times)
print(f"Shape: {arr.shape}")
print(f"Size: {arr.size}")
print(f"Data type: {arr.dtype}")

lap_times_in_min=arr/60
print(lap_times_in_min)
```

**4.** Create:  
* An array of zeros of shape (3,3)  
* An array of ones of shape (2,5)  
* An array filled with 7  



```python
import numpy as np

arr_zeros=np.zeros((3,3))
arr_ones=np.ones((2,5))
arr_seven=np.full((3,3),7)
print("array of zeros of shape (3,3): ")
print(arr_zeros)
print("\narray of ones of shape (2,5): ")
print(arr_ones)
print("\narray filled with 7: ")
print(arr_seven)
```

**4.** Create a 4×4 matrix from 0–15. Then, extract:  
* The first row  
* The last column  
* A 2×2 subarray from the center  
</div>

```python
matrix=np.random.randint(0,15,(4,4))
print(matrix)
first_row=matrix[0:1,]
print(f"\nfirst row {first_row}")

last_col=matrix[:,3]
print(f"\nlast column {last_col}")

sub_array=matrix[1:3,1:3]
print("\n2*2 subarray: ")
print(sub_array)
```
