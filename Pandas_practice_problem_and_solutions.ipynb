### **📝 Task 6**

- Perform **Arithmetic operations** on columns: Create `'Total_Value' = Price * Quantity`.
- Find the **total sum**, **average** of `'Price'`, and find the product with the **maximum quantity**.

````python
from numpy import average
data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150]
}
data_frame=pd.DataFrame(data)
data_frame['Total_Value']=data_frame.loc[:,'Price']*data_frame.loc[:,'Quantity']
print(f"data frame:\n{data_frame}\n")
# print()

total_sum=sum(data_frame.loc[:,'Price'])
print(f"Total sum: {total_sum}\n")

total_average=average(data_frame.loc[:,'Price'])
print(f"Total average: {total_average}\n")

# product_max_id=data_frame['Quantity'].idxmax()
print(f"\nProduct with Maxium Quantity: {data_frame.loc[data_frame['Quantity'].idxmax(),'Product']}\n")
````
