## A. The Homework Bot (Easy)

### Problem Statement

The lab built a homework bot. It does not give long speeches. It only answers one question: did this student pass the AI quiz?

Write a function `check_pass(marks)` that **returns** `True` if `marks >= 40`, otherwise `False`. Do not print inside the function. Add a short docstring — even bots deserve documentation.

Then read one quiz mark, call the function, and print `Pass` or `Fail`.

### Input

One line: an integer `marks` (`-100 ≤ marks ≤ 100`).

### Output

Print `Pass` or `Fail`.

### Examples

**Input**
```text
73
```

**Output**
```text
Pass
```

**Input**
```text
39
```

**Output**
```text
Fail
```

````python
def check_pass(marks):
  """
  Check pass or fail
  marks --> int
  """
  if marks >=40:
    return True
  else:
    return False


marks=check_pass(int(input()))
if marks==True:
  print("Pass")
else:
  print("Fail")

````

## B. Three Training Runs (Easy)

### Problem Statement

You trained a tiny model three times. Same data, three different random seeds. The accuracies came out different — models have moods, like you before coffee.

Write `average(a, b, c)` that **returns** the mean accuracy as a float:

`(a + b + c) / 3`

Call it with **positional arguments** and print only the returned value. The report needs a decimal, so the supervisor believes you actually divided.

### Input

One line: three numbers `a b c` (the three accuracies).

### Output

Print the average as a float (for example, `20.0`, not `20`).

### Example

**Input**
```text
10 20 30
```

**Output**
```text
20.0
```

````python
def average(a,b,c):
  return (a+b+c)/3

a,b,c=map(int,input().split())

print(average(a,b,c))
````

## D. The Model Registry (Medium)

### Problem Statement

The lab keeps a registry of models. If nobody mentions a learning rate, the senior's rule is: **use `0.01`**. "If you forget `lr`, the default will remember for you."

Write `model_info(name, layers, lr=0.01)` that **returns**:

`Model: <name>, Layers: <layers>, LR: <lr>`

Today there are `t` registration requests.

- **Type 1:** The intern is lazy: only `name` and `layers`. Use the default `lr`. Call with positional arguments.
- **Type 2:** The intern typed fields in a weird order. Call with **keyword arguments** (`layers=...`, `name=...`).
- **Type 3:** The intern is brave and sets a custom learning rate.

Print the registry line for each request.

### Input

First line: integer `t` (`1 ≤ t ≤ 10`).

Each of the next `t` lines:

- `1 name layers`
- `2 name layers`
- `3 name layers lr`

### Output

Print `t` registry lines.

### Example

**Input**
```text
3
1 TinyNet 3
2 ResNet 5
3 FastNet 2 0.001
```

**Output**
```text
Model: TinyNet, Layers: 3, LR: 0.01
Model: ResNet, Layers: 5, LR: 0.01
Model: FastNet, Layers: 2, LR: 0.001
```

````python
def model_info(name,layers,lr=.01):
  print(f"Model: {name} , Layers: {layers} , LR: {lr}")

t=int(input())
for i in range(t):
  info=list(input().split())
  if info[0]=="1":
    model_info(info[1],info[2])
  elif info[0]=="2":
    model_info(layers=info[2],name=info[1])
  else:
    model_info(info[1],info[2],info[3])
````

## E. Clean Images, Then Features (Medium)

### Problem Statement

You have a folder of face-image IDs for an age-detection toy model. Odd IDs are blurry selfies. Even IDs are clean.

Keep only the clean (even) IDs, then turn each one into a dummy feature by squaring it. Yes, squaring an ID is a silly feature. The senior said: "First learn filter and map. Fancy embeddings can wait."

You must use `filter()` + `lambda` to keep evens, then `map()` + `lambda` to square. Print the final list of features.

### Input

First line: integer `n` (`1 ≤ n ≤ 100`) — the number of IDs.

Second line: `n` integers — the image IDs.

### Output

Print a Python list of squared even IDs, in the same order.

Example: `[4, 16, 36]`

### Example

**Input**
```text
6
1 2 3 4 5 6
```

**Output**
```text
[4, 16, 36]
```

**Hint:** First remove the blurry faces (`filter`), then create the features (`map`). If you reverse the order, the square of `1` will also sneak in — and the dataset will be ruined.

````python
n=int(input())
numbers=list(map(int,input().split()))
even=filter(lambda x:x%2==0 , numbers)
res=list(map(lambda x:x**2,even))
print(res)
````

## F. GPU Can Eat Only Three Images (Hard)

### Problem Statement

TinyNet is hungry. You have a list of training samples (think: face images for age-group classification). The lab GPU is small. It can swallow only `batch_size` samples at a time.

Write a **generator** `batch_generator(data, batch_size)` that yields one batch (a list) at a time — like giving the GPU one plate, not the whole buffet. The last plate may have fewer samples. If `batch_size < 1`, the GPU is off: yield nothing.

Also write `batch_mean(batch)` that **returns** the average of the numbers in that batch. The senior uses this as a tiny sanity check before real training.

Read the samples and the batch size. For each batch, print:

`[1, 2, 3] mean=2.0`

### Input

First line: two integers `n` and `batch_size` (`n ≥ 1`).

Second line: `n` integers — the sample values.

### Output

Print one line per batch: the batch list, a space, then `mean=` and the mean.

If `batch_size < 1`, print nothing.

### Example

**Input**
```text
8 3
1 2 3 4 5 6 7 8
```

**Output**
```text
[1, 2, 3] mean=2.0
[4, 5, 6] mean=5.0
[7, 8] mean=7.5
```

````python
import math
def batch_generator(data,batch_size):
  n=math.ceil(len(data)/batch_size)
  p=0
  b=batch_size
  for i in range(n):
    ans=(sum(data[p:b]))/len(data[p:b])
    print(f"{data[p:b]}  mean={ans}")
    p=b
    b=p+batch_size

n,batch_size=map(int,input().split())
data=list(map(int,input().split()))
if batch_size<1:
  print("Nothing")
else:
  batch_generator(data,batch_size)
````

````python

def batch_generator(data,batch_size):
  if batch_size < 1:
    return

  for i in range(0,len(data),batch_size):
    yield data[i:i+batch_size]

def batch_mean(batch):
  return (sum(batch)/len(batch))

n,batch_size=map(int,input().split())
data=list(map(int,input().split()))

for batch in batch_generator(data,batch_size):
  print(f"{batch}  mean={batch_mean(batch)}")
````
